# 08 · Cosmology Scalings & Densities (Formula‑First, with units)Critical density and component scalings:$$ \rho_c=\frac{3H_0^2}{8\pi G},\quad \rho_{m,0}=\Omega_{m,0}\rho_c,\quad \rho_{\Lambda,0}=\Omega_{\Lambda,0}\rho_c. $$Evolution with redshift: $\rho_m\propto (1+z)^3$, $\rho_\Lambda=\text{const}$, and $n_\gamma\propto T^3$ with $T(z)=T_0(1+z)$.

## When to use- Compare matter vs dark-energy dominance with z- Scale number densities with temperature

---## Example A — Present-day densities and ratio at high z**Symbols & units**| Symbol | Meaning | Units ||---|---|---|| $H_0$ | Hubble constant | km·s$^{-1}$·Mpc$^{-1}$ || $\Omega_{m,0}$, $\Omega_{\Lambda,0}$ | density parameters | — || $\rho_c$ | critical density | kg·m$^{-3}$ || $z$ | redshift | — |

```python
import numpy as npfrom astropy import units as u, constants as constdef rho_crit(H0): return (3*(H0**2)/(8*np.pi*const.G)).to(u.kg/u.m**3)H0 = 70*u.km/u.s/u.Mpc; Om0, Ol0 = 0.3, 0.7rc = rho_crit(H0); rho_m0 = Om0*rc; rho_L0 = Ol0*rcz = 7.0; ratio = (rho_m0*(1+z)**3)/rho_L0rc, rho_m0, rho_L0, ratio.decompose()
```
