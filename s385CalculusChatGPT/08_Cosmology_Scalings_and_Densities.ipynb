# 08 · Cosmology Scalings and Densities (with units)## DefinitionCritical density is $$ \rho_c = \frac{3H_0^2}{8\pi G}. $$Today’s matter and dark-energy densities are $\rho_{m,0}=\Omega_{m,0}\rho_c$ and $\rho_{\Lambda,0}=\Omega_{\Lambda,0}\rho_c$.With redshift: $\rho_m(z)=\rho_{m,0}(1+z)^3$ while $\rho_\Lambda$ is constant. Photon number density scales as $n_\gamma\propto T^3$ with $T(z)=T_0(1+z)$.

## Applications- Determine when matter or dark energy dominates.- Compare photon-to-baryon ratios if $T_0$ differs but baryon density stays the same.

## Key formulas and concepts- Ratio at redshift $z$: $\frac{\rho_m}{\rho_\Lambda} = \frac{\Omega_{m,0}}{\Omega_{\Lambda,0}}(1+z)^3$.- $n_\gamma$ scales as $T^3$; if $T_0$ halves, $n_\gamma$ drops by $1/8$.
