# 11 · Gradients and Multiple Integrals (with units)## DefinitionThe gradient of a scalar field $\rho(x,y,z)$ is the vector of partial derivatives:$$ \nabla\rho = \left(\frac{\partial\rho}{\partial x},\frac{\partial\rho}{\partial y},\frac{\partial\rho}{\partial z}\right), $$which points in the direction of steepest increase. Total mass in a spherically symmetric distribution is$$ M = 4\pi\int_0^{R} \rho(r) r^2\,dr. $$

## Applications- Interpret how fast density, temperature, or potential changes in space.- Compute total mass given a radial density profile.

## Key formulas and concepts- In spherical symmetry: $dV = 4\pi r^2 dr$.- For an exponential profile $\rho(r)=\rho_0 e^{-r/R}$ the integral is analytic, but we can also check with SymPy.

```python
import sympy as sprho0, R, r, Rstar = sp.symbols('rho0 R r Rstar', positive=True)rho = rho0*sp.exp(-r/R)M = sp.integrate(4*sp.pi * r**2 * rho, (r, 0, Rstar))grad_demo = sp.Matrix([sp.diff(rho0*(1+0.1*sp.Symbol('x')-0.2*sp.Symbol('y')), v) for v in (sp.Symbol('x'), sp.Symbol('y'))])M, grad_demo
```
