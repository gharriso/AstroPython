# 11 · Gradients & Multiple Integrals (Formula‑First, with units)The **gradient** of a scalar field ρ(x,y,z) is$$ \nabla\rho = \left(\frac{\partial\rho}{\partial x},\frac{\partial\rho}{\partial y},\frac{\partial\rho}{\partial z}\right) $$and points in the direction of steepest increase. In spherical symmetry, a total mass is$$ M=4\pi\int_0^R \rho(r) r^2\,dr. $$

## When to use- Interpret how quickly fields change in space- Compute totals from spherically symmetric profiles

---## Example A — Gradient of a toy 2D field**Model:** $\rho(x,y)=\rho_0(1+0.1x-0.2y)$

```python
import sympy as sprho0, x, y = sp.symbols('rho0 x y')rho = rho0*(1 + 0.1*x - 0.2*y)grad = [sp.diff(rho, v) for v in (x,y)]grad
```

---## Example B — Mass of an exponential sphere**Model:** $\rho(r)=\rho_0 e^{-r/R}$

```python
import sympy as sprho0, r, R = sp.symbols('rho0 r R', positive=True)rho = rho0*sp.exp(-r/R)M = sp.integrate(4*sp.pi*r**2*rho, (r, 0, 10*R))sp.simplify(M)
```
