# 00 · Basics of Differentiation (with units)## DefinitionDifferentiation measures how quickly one quantity changes with respect to another. If a function is written as $y = f(x)$, the derivative is the instantaneous rate of change:$$ f'(x) = \frac{dy}{dx} = \lim_{\Delta x \to 0} \frac{f(x + \Delta x) - f(x)}{\Delta x}. $$- Prime notation: $f'(x)$ and $f''(x)$.- Leibniz notation: $\frac{dy}{dx}$ and $\frac{d^2y}{dx^2}$.- Time derivatives: $\dot{x} = \frac{dx}{dt}$ and $\ddot{x} = \frac{d^2x}{dt^2}$.In physics, the meanings are concrete:- Velocity is the derivative of position with respect to time: $v(t) = \frac{dr}{dt}$.- Acceleration is the derivative of velocity: $a(t) = \frac{dv}{dt}$.- In cosmology, the Hubble parameter is $H(t) = \frac{\dot a}{a}$ where $a(t)$ is the scale factor.

## When to use differentiation- You have a quantity as a function and want its rate (slope) at any point.- You want to convert between related rates (position → velocity → acceleration, flux → spectral slope).- You need to optimise (max/min) or analyse how fast curves rise or fall.**Units:** the units of $\frac{dy}{dx}$ are "units of $y$ divided by units of $x$". If $y$ is metres and $x$ is seconds, then $\frac{dy}{dx}$ is m·s$^{-1}$.

## Visual pictureThe derivative at $x=x_0$ is the slope of the tangent line to the curve at that point. The sign indicates the direction of change; the magnitude tells how steep it is.
