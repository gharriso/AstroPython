# 04 · Integration and Areas (with units)## DefinitionIntegration turns a rate into a total. In astronomy we use it to compute distances from velocities, luminosities from flux distributions, and mass from density profiles. When analytic antiderivatives are awkward, we integrate numerically.

## Applications- Distance from velocity: $s=\int v\,dt$.- Luminosity from flux density over wavelength: $L=4\pi d^2\int F_\lambda\,d\lambda$.- Mass from density: $M=\int\rho\,dV$; spherical: $M=4\pi\int_0^R\rho(r)r^2\,dr$.

## Key formulas and concepts- Indefinite: $\int f(x)\,dx$; Definite: $\int_a^b f(x)\,dx$.- Numeric trapezoid rule: `np.trapz(y, x)`; spacing must match units.
