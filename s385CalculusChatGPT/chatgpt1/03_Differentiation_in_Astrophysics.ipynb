# 03 · Differentiation in Astrophysics (with units)## DefinitionDifferentiation links physical quantities that change with time or position. In astronomy we routinely compute:- $v(t)=\frac{dr}{dt}$ to get velocities from positions.- $a(t)=\frac{dv}{dt}$ to measure accelerations.- Spectral slopes or temperature gradients, e.g. $\frac{dT}{dr}$.We use SymPy for exact derivatives and then evaluate with numbers and units using Astropy.

## Applications- Orbital geometry: $r(\theta)$ and $\frac{dr}{d\theta}$ for orbits.- Radiation: $u(T)=aT^4 \Rightarrow \frac{du}{dT}=4aT^3$.- Growth laws: if $M(t)=M_0 e^{ft}$ then $\frac{dM}{dt}=fM$.

## Key formulas and concepts- Prime and Leibniz notations are interchangeable: $f'(x)=\frac{df}{dx}$.- Chain-like derivatives are handled by SymPy automatically.- Units follow the variable you differentiate with respect to (e.g., per radian, per second).
