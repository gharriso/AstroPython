# 01 · Basics of Integration (with units)## DefinitionIntegration accumulates small contributions to obtain a total. If $v(t)$ is a speed, the distance travelled from $t=a$ to $t=b$ is$$ s = \int_a^b v(t)\,dt. $$Indefinite integrals return a function whose derivative is the original integrand:$$ \int f(x)\,dx = F(x) + C, \quad \text{with } F'(x)=f(x). $$- Leibniz: $\int f(x)\,dx$ and $\int_a^b f(x)\,dx$.- Python: `integrate(f, x)` and `integrate(f, (x, a, b))`.

## When to use integration- Convert a rate into a total (velocity → distance, power → energy, density → mass).- Compute areas under curves.- Solve differential equations (by integrating a rate function).**Units:** the units of an integral are "units of the integrand times units of the integration variable". Example: m·s$^{-1}$ times s equals m.

## Example with units: luminosity from a line profileIf $F_\lambda(\lambda)$ is the flux density at Earth and $d$ is distance, then a band-limited luminosity is$$ L \approx 4\pi d^2 \int_{\lambda_1}^{\lambda_2} F_\lambda(\lambda)\,d\lambda. $$
