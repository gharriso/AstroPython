# 05 · Numerical Integration in Cosmology (with units)## DefinitionIn a flat $\Lambda$CDM toy model, the Hubble parameter is $H(z)=H_0\sqrt{\Omega_m(1+z)^3+\Omega_\Lambda}$. The comoving distance is$$ d_c(z) = c \int_0^z \frac{dz'}{H(z')} $$and the lookback time is$$ t_{LB}(z) = \int_0^z \frac{dz'}{(1+z')H(z')}. $$These are evaluated numerically with `np.trapz`.

## Applications- Convert redshift to distances for lensing and distance modulus.- Estimate cosmic ages and lookback times.

## Key formulas and concepts- $d_A = d_c/(1+z)$ and $d_L = d_c(1+z)$ (flat universe).- Keep consistent units: km s$^{-1}$ Mpc$^{-1}$, Mpc, and $c$ in km s$^{-1}$.

```python
import numpy as npimport matplotlib.pyplot as pltfrom astropy import units as ufrom astropy import constants as constc = const.c.to(u.km/u.s)H0 = 70 * u.km/u.s/u.MpcOm0, Ol0 = 0.3, 0.7def H_of_z(z):    return H0*np.sqrt(Om0*(1+z)**3 + Ol0)def comoving_distance(z, n=40000):    zz = np.linspace(0, z, n)    integrand = (1/H_of_z(zz)).to(u.Mpc/(u.km/u.s)).value    val = np.trapz(integrand, zz) * (c.value)    return val * u.Mpcdef lookback_time(z, n=60000):    zz = np.linspace(0, z, n)    integrand = (1/((1+zz)*H_of_z(zz))).to(u.Mpc/(u.km/u.s)).value    sec = np.trapz(integrand, zz) * (u.Mpc.to(u.km))    return (sec * u.s).to(u.Gyr)zs = np.linspace(0, 8, 120)dcs = [comoving_distance(z).to_value(u.Mpc) for z in zs]plt.figure()plt.plot(zs, dcs)plt.xlabel("z")plt.ylabel("d_c (Mpc)")plt.title("Comoving distance vs z")plt.show()lookback_time(2.0)
```

## Exercises1. Compute $d_c$ for $H_0=43$ km s$^{-1}$ Mpc$^{-1}$, $\Omega_m=0.05$, $\Omega_\Lambda=0.95$ at $z=7$.2. Plot $t_{LB}(z)$ from 0 to 10.3. Using your $d_c$, compute and plot $d_A$ and $d_L$ vs $z$.

## Solutions

```python
import numpy as npimport matplotlib.pyplot as pltfrom astropy import units as ufrom astropy import constants as constc = const.c.to(u.km/u.s)H0 = 43 * u.km/u.s/u.MpcOm0, Ol0 = 0.05, 0.95def H_of_z(z):    return H0*np.sqrt(Om0*(1+z)**3 + Ol0)def comoving_distance(z, n=50000):    zz = np.linspace(0, z, n)    integrand = (1/H_of_z(zz)).to(u.Mpc/(u.km/u.s)).value    val = np.trapz(integrand, zz) * (c.value)    return val * u.Mpcdef lookback_time(z, n=60000):    zz = np.linspace(0, z, n)    integrand = (1/((1+zz)*H_of_z(zz))).to(u.Mpc/(u.km/u.s)).value    sec = np.trapz(integrand, zz) * (u.Mpc.to(u.km))    return (sec * u.s).to(u.Gyr)print("d_c(z=7) =", comoving_distance(7))zgrid = np.linspace(0, 10, 120)tLB = [lookback_time(z).value for z in zgrid]plt.figure(); plt.plot(zgrid, tLB); plt.xlabel("z"); plt.ylabel("t_LB (Gyr)"); plt.title("Lookback time"); plt.show()zgrid = np.linspace(0.01, 6, 120)dc = np.array([comoving_distance(z).to_value(u.Mpc) for z in zgrid])dA = dc/(1+zgrid); dL = dc*(1+zgrid)plt.figure(); plt.plot(zgrid, dA); plt.xlabel("z"); plt.ylabel("d_A (Mpc)"); plt.title("Angular-diameter distance"); plt.show()plt.figure(); plt.plot(zgrid, dL); plt.xlabel("z"); plt.ylabel("d_L (Mpc)"); plt.title("Luminosity distance"); plt.show()
```
