# 10 · Distance Modulus and Relativistic Jets (with units)## DefinitionDistance modulus relates apparent and absolute magnitudes:$$ \mu = m - M = 5\log_{10}(d_L/10\,\text{pc}), \quad d_L = 10^{(\mu/5+1)}\ \text{pc}. $$Apparent superluminal motion arises from light-travel-time effects in relativistic jets. A simple observer-time estimate is$$ v_{app} \approx \mu_\theta\, d_A, \quad \beta_{app}=v_{app}/c, \quad \mu_\theta = \Delta\theta/\Delta t_{obs}. $$Some texts include an extra $(1+z)$ factor when converting to emitter-frame time. We show both numbers.

## Applications- Convert $m$ and $M$ to a luminosity distance.- Estimate jet apparent speed from a measured proper motion and $d_A$.

## Key formulas and concepts- $d_A = d_L/(1+z)^2$ (flat case).- $\beta_{app}$ peaks near $\cos\theta \approx \beta$; a handy lower bound on $\Gamma$ is $\Gamma \gtrsim \sqrt{1+\beta_{app}^2}$.
