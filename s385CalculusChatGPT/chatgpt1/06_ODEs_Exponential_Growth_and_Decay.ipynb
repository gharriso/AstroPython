# 06 · ODEs — Exponential Growth and Decay (with units)## DefinitionFor $\dot y = k y$ with constant $k$, the solution is exponential: $y(t)=y_0 e^{kt}$. In astrophysics this models black-hole mass growth at a constant Eddington fraction, or simple decays/cooling with constant fractional rates.

## Applications- Black-hole growth: $\dot M = f M$.- Radioactive-like decays: $\dot N = -\lambda N$.- E-folding and doubling times: $t_{\times 2}=\frac{\ln 2}{k}$, $t_{+p\%}=\frac{\ln(1+p/100)}{k}$.

## Key formulas and concepts- General solution: $y(t)=y_0 e^{kt}$.- Growth by 50% time: $t_{50\%}=\frac{\ln 1.5}{k}$.
