# 05 · Numerical Integration in Cosmology (Formula‑First, with units)In a flat ΛCDM toy model,$$ H(z) = H_0\sqrt{\Omega_m(1+z)^3 + \Omega_\Lambda}. $$**Comoving distance:** $$ d_c(z) = c\int_0^z \frac{dz'}{H(z')}. $$**Lookback time:** $$ t_{LB}(z) = \int_0^z \frac{dz'}{(1+z')H(z')}. $$We evaluate these with numerical integration (`np.trapz`) and attach units with Astropy.

## When to use- Convert redshift to distances and ages- Feed distances to lensing & distance-modulus problems

---## Example A — Distances in flat ΛCDM**Symbols & units**| Symbol | Meaning | Units ||---|---|---|| $H_0$ | Hubble constant today | km·s$^{-1}$·Mpc$^{-1}$ || $\Omega_m,\Omega_\Lambda$ | density parameters | — || $d_c, d_A, d_L$ | distances | Mpc |

```python
import numpy as npfrom astropy import units as u, constants as constc = const.c.to(u.km/u.s)H0 = 70 * u.km/u.s/u.MpcOm, Ol = 0.3, 0.7def H(z): return H0*np.sqrt(Om*(1+z)**3 + Ol)def comoving(z, n=40000):    zz = np.linspace(0, z, n)    integrand = (1/H(zz)).to(u.Mpc/(u.km/u.s)).value    val = np.trapz(integrand, zz) * c.value    return val * u.Mpcdef lookback(z, n=40000):    zz = np.linspace(0, z, n)    integrand = (1/((1+zz)*H(zz))).to(u.Mpc/(u.km/u.s)).value    sec = np.trapz(integrand, zz) * (u.Mpc.to(u.km))    return (sec * u.s).to(u.Gyr)z = 2.0dc = comoving(z)dA = dc/(1+z); dL = dc*(1+z)dc.to(u.Mpc), dA.to(u.Mpc), dL.to(u.Mpc), lookback(z)
```
