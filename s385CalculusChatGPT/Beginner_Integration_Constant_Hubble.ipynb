
# Beginner Integration: Constant Hubble Expansion

We’ll use a very simple cosmology toy model as a **first integration example**:

$$
\frac{dD}{dt} = H_0\,D(t),
$$

where
- $D(t)$ is the proper distance to a comoving galaxy, and  
- $H_0$ is taken as a **constant** (toy assumption to keep the calculus easy).

This leads to **exponential growth**, and is perfect for practicing separation of variables.


```python

# Imports
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing(use_latex=True)
```


## 1) Solve by separation of variables

Given
$$
\frac{dD}{dt} = H_0 D,
$$

separate variables and integrate:
$$
\int \frac{1}{D}\,dD = \int H_0\,dt
\quad\Rightarrow\quad
\ln D = H_0 t + C.
$$

Exponentiate both sides:
$$
D(t) = C'\,e^{H_0 t}.
$$

If we set the initial condition $D(0)=D_0$, then $C'=D_0$ and
$$
\boxed{\;D(t) = D_0\,e^{H_0 t}\;}
$$


```python

# Symbolic verification with SymPy
t, H0 = sp.symbols('t H0', positive=True, real=True)
D = sp.Function('D')

ode = sp.Eq(sp.diff(D(t), t), H0*D(t))
solution = sp.dsolve(ode)      # general solution
solution
```


The `dsolve` result is equivalent to
$$
D(t)=C_1 e^{H_0 t}.
$$
Applying $D(0)=D_0$ gives $C_1=D_0$.


```python

D0 = sp.symbols('D0', positive=True, real=True)
specific = sp.Eq(D(t), D0*sp.exp(H0*t))
specific
```


## 2) Units and a 5 Gyr example

A commonly quoted value is
$$
H_0 \approx 70\;\text{km s}^{-1}\,\text{Mpc}^{-1}.
$$

For time in **seconds**, convert $H_0$ to s$^{-1}$. For long spans (Gyr), you can also convert to Gyr$^{-1}$.

We’ll compute the **growth factor** over $\Delta t=5\,\text{Gyr}$:
$$
\text{growth} = \frac{D(t_0+\Delta t)}{D(t_0)} = e^{H_0\,\Delta t}.
$$


```python

# Numerical example: growth over 5 Gyr
H0_km_s_Mpc = 70.0

# Unit conversions
Mpc_m = 3.085677581491367e22      # meters
H0_val = H0_km_s_Mpc * 1000.0 / Mpc_m # [1/s]

sec_per_year = 3.15576e7
Gyr_s = 1e9 * sec_per_year
Delta_t = 5.0 * Gyr_s             # 5 Gyr in seconds

growth = float(sp.exp(H0_val * Delta_t))
growth
```


The number printed is the **distance multiplication factor** over 5 Gyr for this toy model.
If a galaxy is $D_0=1000\,\text{Mpc}$ away today, then after 5 Gyr:
$$
D = D_0\times \text{growth}.
$$



## 3) Plot $D(t)$ and the growth factor

We’ll plot the ratio $D(t)/D_0 = e^{H_0 t}$ over a 0–5 Gyr window.


```python

# Create a time array (0..5 Gyr) and plot D(t)/D0
ts = np.linspace(0.0, 5.0, 200)           # Gyr
Dt_over_D0 = np.exp(H0_val * ts * Gyr_s)  # e^{H0 t}

plt.figure(figsize=(6,4))
plt.plot(ts, Dt_over_D0)
plt.xlabel("t  [Gyr]")
plt.ylabel("D(t) / D0")
plt.title("Constant-Hubble toy model: distance growth")
plt.grid(True)
plt.show()

Dt_over_D0[-1]  # growth at 5 Gyr (matches 'growth' above)
```
