# 04 · Integration and Areas (Formula‑First, with units)Integration turns a rate into a total. We will combine symbolic antiderivatives and numeric `np.trapz` checks, always starting with the physical formula and a units table.

## When to use- Convert instantaneous rates into totals (distance, energy)- Accumulate density over volume to obtain mass- Compute areas under curves for interpretation

---## Example A — From acceleration to velocity & distance**Kinematics formulas:**$$ v(t)=\int a(t)\,dt+v(t_0), \qquad s(t)=\int v(t)\,dt + s(t_0). $$**Model:** $a(t)=a_0 - kt$**Symbols & units**| Symbol | Meaning | Units ||---|---|---|| $a(t)$ | acceleration | m·s$^{-2}$ || $v(t)$ | velocity | m·s$^{-1}$ || $s(t)$ | displacement | m || $k$ | constant | s$^{-2}$ |

```python
import sympy as spfrom astropy import units as ut = sp.symbols('t', nonnegative=True)a0, k = 4.0, 0.2a = a0 - k*tv = sp.integrate(a, t) + 0  # v0=0s = sp.integrate(v, t) + 0  # s0=0v, s
```

---## Example C — Mass from density (spherical shell)**Formula:** $$ M(R) = 4\pi \int_0^R \rho(r) r^2\,dr. $$**Model:** $\rho(r)=\rho_0 e^{-r/R_0}$

```python
import sympy as spfrom astropy import units as urho0, r, R, R0 = sp.symbols('rho0 r R R0', positive=True)rho = rho0*sp.exp(-r/R0)M = sp.integrate(4*sp.pi*r**2*rho, (r, 0, R))sp.simplify(M)
```
