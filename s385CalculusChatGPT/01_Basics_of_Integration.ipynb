# 01 · Basics of Integration (Formula‑First, Astrophysics Context)Integration accumulates small contributions to obtain a total. We’ll start each example with a **full formula**, define symbols and units, then compute in Python (symbolically or numerically) with unit‑safety.

## 1. What is Integration?For a rate $f(x)$, the total accumulated amount from $x=a$ to $x=b$ is the **definite integral**:$$ \int_a^b f(x)\,dx. $$An **indefinite integral** returns an antiderivative $F(x)$ such that $F'(x)=f(x)$:$$ \int f(x)\,dx = F(x) + C. $$**Units:** (units of integrand) × (units of the integration variable).

## 2. When do astronomers integrate?- Distance from velocity: $s=\int v\,dt$.- Luminosity from a flux density profile: $L=4\pi d^2\int F_\lambda\,d\lambda$.- Mass from density: $M=\int \rho\,dV$ (spherical: $4\pi\int_0^R \rho(r)r^2\,dr$).- Cosmological distances and ages: $d_c(z)=c\int_0^z dz'/H(z')$.

```python
import sympy as spfrom astropy import units as ur, R = sp.symbols('r R', positive=True)rho0 = 5e-20 * (u.kg/u.m**3)        # constant density for demorho = rho0.value                    # numeric factor to integrateM_expr = 4*sp.pi*sp.integrate(rho * r**2, (r, 0, R)) * u.kgM_at = M_expr.subs({R: (2*u.pc).to(u.m).value}).to(u.kg)M_expr, M_at
```
