# 09 · Gravitational Lensing — Distances & Einstein Mass (Formula‑First, with units)For a point-mass lens, the Einstein angle is$$ \theta_E=\sqrt{\frac{4GM}{c^2}\frac{d_{A,ls}}{d_{A,l}d_{A,s}}}. $$Solve for mass:$$ M=\frac{c^2}{4G}\,\theta_E^2\,\frac{d_{A,l}d_{A,s}}{d_{A,ls}}. $$

## When to use- Estimate mass enclosed within Einstein radius- Explore sensitivity to distances and θ_E

---## Example A — Mass from an Einstein ring**Symbols & units**| Symbol | Meaning | Units ||---|---|---|| $\theta_E$ | Einstein angle | rad (convert arcsec→rad) || $d_{A,l}$, $d_{A,s}$, $d_{A,ls}$ | angular‑diameter distances | Mpc || $M$ | mass | kg or $M_\odot$ |

```python
import numpy as npfrom astropy import units as u, constants as constdef mass_from_theta(theta_arcsec, dA_l, dA_s, dA_ls):    th = (theta_arcsec * u.arcsec).to(u.rad)    M = (const.c**2/(4*const.G)) * (th**2) * (dA_l*dA_s/dA_ls)    return M.to(u.M_sun)dA_l = 900*u.Mpc; dA_s = 1200*u.Mpc; dA_ls=500*u.Mpcmass_from_theta(0.6, dA_l, dA_s, dA_ls)
```

## Exercises1) Sweep θ_E from 0.2″ to 1.0″ and tabulate M at fixed distances. 2) If d_A change by ±10%, how does M shift?

## Solutions

```python
thetas = np.linspace(0.2,1.0,9)Ms = [mass_from_theta(th, dA_l, dA_s, dA_ls) for th in thetas]list(zip(thetas, [m.to_value(u.M_sun) for m in Ms]))
```
