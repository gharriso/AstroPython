# 09 · Gravitational Lensing — Distances and Einstein Mass (with units)## DefinitionFor a point-mass lens, the Einstein angle is$$ \theta_E = \sqrt{\frac{4GM}{c^2}\frac{d_{A,ls}}{d_{A,l}d_{A,s}}}. $$Solving for mass:$$ M = \frac{c^2}{4G}\,\theta_E^2\,\frac{d_{A,l}d_{A,s}}{d_{A,ls}}. $$Distances are angular-diameter distances between lens, source, and observer.

## Applications- Estimate enclosed mass within the Einstein radius from an observed ring/arc.- Explore how $M$ scales with $\theta_E$ and the distance ratio.

## Key formulas and concepts- $\theta_E$ in radians; convert arcseconds: 1 arcsec $=\pi/(648000)$ rad.- Use consistent Mpc↔m conversions.

```python
import numpy as npfrom astropy import units as u, constants as constarcsec_to_rad = np.pi/(180*3600)def mass_from_theta(theta_arcsec, dA_l, dA_s, dA_ls):    th = theta_arcsec*arcsec_to_rad    M = (const.c**2/(4*const.G)) * (th**2) * (dA_l*dA_s/dA_ls)    return M.to(u.M_sun)dA_l = 900 * u.Mpc; dA_s = 1200 * u.Mpc; dA_ls = 500 * u.Mpcmass_from_theta(0.6, dA_l, dA_s, dA_ls)
```
