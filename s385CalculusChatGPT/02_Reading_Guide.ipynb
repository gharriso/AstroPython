# 02 · Reading Guide — How to Use This Workbook SetThis pack is a **self-contained calculus companion** for S384/S385: every example starts with a **full formula**, definitions of symbols with **units**, and a Python implementation (symbolic or numeric) that avoids manual rule‑drilling.

## What you’ll learn- Differentiation and integration as they appear in astronomy and cosmology- When to apply each technique, with unit‑safe code- ODEs of the form $\dot y = k y$, special relativity kinematics, and cosmology distances- Interpreting gradients and performing simple volume integrals

## Python stack- **SymPy** for symbolic derivatives and integrals- **NumPy** for arrays and numerical integration (`np.trapz`)- **Matplotlib** for single plots (no subplots)- **Astropy** for physical constants and `units`All examples attach units, then convert as needed with `.to(new_unit)`.

## Notation cheatsheet- Derivatives: $f'(x)$, $\frac{df}{dx}$, $\dot x = \frac{dx}{dt}$- Integrals: $\int f(x)\,dx$, $\int_a^b f(x)\,dx$- Cosmology distances (flat): $d_A=d_c/(1+z)$, $d_L=d_c(1+z)$, $d_c=c\int dz/H(z)$
