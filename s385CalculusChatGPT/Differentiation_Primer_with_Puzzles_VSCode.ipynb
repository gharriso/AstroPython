
## 1) Principles & Notation

### Derivative as a limit
For a differentiable function $f(x)$,
$$
f'(x) = \lim_{h\to 0} \frac{f(x+h) - f(x)}{h}.
$$

### Notation equivalences
- $f'(x)$ — Lagrange notation  
- $\dfrac{df}{dx},\; \dfrac{dy}{dx}$ — Leibniz notation (if $y=f(x)$)  
- $D_x f,\; Df(x),\; \dfrac{d}{dx}f(x)$ — operator forms

### Common rules (single variable)
- **Constant rule**: $\dfrac{d}{dx}(c)=0$
- **Power rule**: $\dfrac{d}{dx}(x^n)=n x^{n-1}$ (for real $n$)
- **Constant multiple**: $\dfrac{d}{dx}(c\,f)=c f'$
- **Sum/Difference**: $\dfrac{d}{dx}(f \pm g)=f' \pm g'$
- **Product rule**: $\dfrac{d}{dx}(f g)= f' g + f g'$
- **Quotient rule**: $\dfrac{d}{dx}\!\left(\frac{f}{g}\right)=\dfrac{f'g - fg'}{g^2}$ (where $g\neq 0$)
- **Chain rule**: $\dfrac{d}{dx} \, f(g(x)) = f'(g(x))\, g'(x)$

### Common derivatives
$$
\begin{aligned}
\frac{d}{dx}\sin x &= \cos x, & \frac{d}{dx}\cos x &= -\sin x, & \frac{d}{dx}\tan x &= \sec^2 x,\\
\frac{d}{dx}e^x &= e^x, & \frac{d}{dx}\ln x &= \frac{1}{x} \;(x>0), & \frac{d}{dx}a^x &= a^x \ln a.
\end{aligned}
$$

### Units reminder
If $y$ has units $U_y$ and $x$ has units $U_x$, then $\dfrac{dy}{dx}$ has units $U_y/U_x$.



## 2) Simple Applications

### (A) Tangent line at a point
Given $y=f(x)$, at $x=a$: slope $m=f'(a)$. Tangent line:
$$
y \approx f(a) + f'(a)(x-a).
$$

Below: compute a tangent line for a sample function and visualize the formula value.


```python

# Tangent line example
x = sp.symbols('x', real=True)
f = sp.exp(x) * sp.cos(x)       # example function
a = sp.Rational(1,2)            # point a = 1/2

slope = sp.diff(f, x).subs(x, a)
fa = f.subs(x, a)
tangent = fa + slope*(x - a)

f, sp.simplify(sp.diff(f, x)), slope, tangent
```


### (B) Simple related rates (one variable proxy)
Suppose $s(t)=\tfrac12 a t^2$ (constant acceleration model). Then $v(t)=s'(t)=a t$.  
Units: if $s$ is metres and $t$ seconds, $v$ is m·s$^{-1}$, $a$ is m·s$^{-2}$.


```python

t, a = sp.symbols('t a', positive=True)
s = sp.Rational(1,2)*a*t**2
v = sp.diff(s, t)
a_units_demo = sp.diff(v, t)  # should be a

s, v, a_units_demo
```


### (C) Optimization (first-derivative test)
Find critical points by solving $f'(x)=0$. Use sign changes or $f''(x)$ to classify.


```python

# Optimization example
x = sp.symbols('x', real=True)
g = x**3 - 3*x**2 + 2
g1 = sp.diff(g, x)
crit = sp.solve(sp.Eq(g1, 0), x)
g2 = sp.diff(g1, x)
crit, [g2.subs(x, c) for c in crit], [g.subs(x,c) for c in crit]
```

## 4) Solutions (in Python)

> Try not to peek until you've attempted the puzzles!


```python

# Puzzle 1
x = sp.symbols('x', real=True)
h = sp.log(sp.sqrt(1 + 3*x**2))
hprime = sp.simplify(sp.diff(h, x))
hprime_at_1 = sp.simplify(hprime.subs(x, 1))
hprime, hprime_at_1
```

```python

# Puzzle 2
x = sp.symbols('x', real=True)
p = sp.exp(2*x)*sp.sin(3*x)
pprime = sp.diff(p, x)

# Solve p'(x) = 0, smallest positive solution
candidates = set()
for guess in [0.05, 0.1, 0.2, 0.3, 0.5, 1.0]:
    try:
        r = float(sp.nsolve(sp.Eq(pprime, 0), guess))
        if r > 0:
            candidates.add(round(r, 12))
    except Exception:
        pass
smallest_pos = min(candidates) if candidates else None
pprime, smallest_pos
```

```python

# Puzzle 3
x = sp.symbols('x', real=True)
q = (x**2 + 1)/(x - 1)
qprime = sp.simplify(sp.diff(q, x))
a = 2
slope = qprime.subs(x, a)
ya = q.subs(x, a)
tangent = sp.simplify(ya + slope*(x - a))
qprime, slope, ya, tangent
```

```python

# Puzzle 4
x = sp.symbols('x', real=True)
f = x*sp.exp(-x)
fprime = sp.diff(f, x)
solution = sp.solve(sp.Eq(fprime, -sp.E**-1), x)
fprime, solution
```

```python

# Puzzle 5
x = sp.symbols('x', positive=True)
P = 20
# rectangle with sides x and y, perimeter 2(x+y)=P => y = P/2 - x
y = P/2 - x
A = sp.simplify(x*y)
A1 = sp.diff(A, x)
x_star = sp.solve(sp.Eq(A1, 0), x)[0]
A_max = A.subs(x, x_star)
x_star, y.subs(x, x_star), sp.simplify(A_max)
```

```python

# Puzzle 6
t = sp.symbols('t', real=True)
s = t**3 - 6*t**2 + 9*t
v = sp.diff(s, t)
a = sp.diff(v, t)

# Direction changes when v(t)=0 and sign changes around the root(s)
roots_v = sp.solve(sp.Eq(v, 0), t)
sign_changes = []
for r in roots_v:
    left = v.subs(t, r-0.1)
    right = v.subs(t, r+0.1)
    sign_changes.append((r, sp.sign(left), sp.sign(right)))

# Speed increasing when v and a have same sign (v*a > 0)
intervals_speed_up = sp.solve_univariate_inequality(v*a > 0, t)

s, v, a, roots_v, sign_changes, intervals_speed_up
```

```python

# Puzzle 7
x, y = sp.symbols('x y', real=True)
F = x**2 + x*y + y**2 - 7
# Implicit derivative dy/dx = -F_x / F_y
Fx = sp.diff(F, x)
Fy = sp.diff(F, y)
dydx = -Fx/Fy

val = sp.simplify(dydx.subs({x:1, y:2}))
Fx, Fy, sp.simplify(dydx), val
```

```python

# Puzzle 8
x = sp.symbols('x', positive=True)
r = (x**2 + 1)**x
# Log differentiation: ln r = x ln(x^2+1)
# r'/r = d/dx [x ln(x^2+1)] = ln(x^2+1) + x * (2x/(x^2+1))
rprime = sp.simplify(sp.diff(r, x))
# Optional: express as r * (...) explicitly
rprime_factor = sp.simplify(rprime / r)
r, rprime, rprime_factor
```

```python

# Puzzle 9
a, b, adot, bdot = sp.symbols('a b adot bdot', real=True)
c = sp.sqrt(a**2 + b**2)
cdot = sp.simplify((a*adot + b*bdot)/c)

# Plug given values
val = cdot.subs({a:3, b:4, adot:2, bdot:-1})
c, cdot, val
```

```python

# Puzzle 10
x = sp.symbols('x', real=True)
u = sp.log(1 + x)
a0 = 0
m = sp.diff(u, x).subs(x, a0)
ua = u.subs(x, a0)
tangent = sp.simplify(ua + m*(x - a0))

approx = tangent.subs(x, sp.Rational(1,10))  # x=0.1
exact = sp.log(1 + sp.Rational(1,10))
error = sp.Abs(exact - approx).evalf()
tangent, approx.evalf(), exact.evalf(), error
```
