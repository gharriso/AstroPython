# 07 · Special Relativity — Lorentz & Simultaneity (Formula‑First)Lorentz transforms relate $(t,x)$ in frames moving at speed $v$:$$ t' = \gamma\left(t - \frac{v x}{c^2}\right),\quad x' = \gamma(x - v t),\quad \gamma=1/\sqrt{1-\beta^2},\;\beta=v/c. $$**Causality test:** $(\Delta ct)^2-(\Delta x)^2\ge 0$ with $\Delta t>0$.**Simultaneity condition:** choose $v$ so $\Delta t'=0$ ⇒ $\beta=\Delta(ct)/\Delta x$ (in $c=1$ units).

## When to use- Check whether one event can influence another- Find the boost that makes two events simultaneous

---## Example A — Exam replica: S385 Q1 (causality & simultaneity)**Events:** A=(3.0,2.0), B=(9.0,0.5), C=(5.5,5.0), D=(10.0,2.5) in $(ct,x)$ units (so $c=1$).**Tasks:** (i) Which of A or B can causally influence D? (ii) Which event can be simultaneous with A in some inertial frame? (iii) What $v$ is required?

```python
from math import sqrtevents = {"A": (3.0,2.0), "B": (9.0,0.5), "C": (5.5,5.0), "D": (10.0,2.5)}def minkowski(E, D="D"):    dct = events[D][0]-events[E][0]; dx = events[D][1]-events[E][1]; s2 = dct**2 - dx**2    return dct, dx, s2for E in ["A","B"]:    dct, dx, s2 = minkowski(E)    print(E, "-> D", "causal?", (dct>0 and s2>=0), "Δct,Δx,s^2=", dct, dx, s2)def beta_for_sim(E1, E2):    dct = events[E2][0]-events[E1][0]; dx = events[E2][1]-events[E1][1]    return float("inf") if abs(dx)<1e-12 else dct/dxfor lab in ["B","C","D"]:    print("β for A &", lab, "simultaneous =", beta_for_sim("A", lab))
```
