# 07 · Special Relativity — Lorentz and Simultaneity## DefinitionLorentz transformations relate time and position in two inertial frames moving at speed $v$ along $x$:$$ t' = \gamma\left(t - \frac{v x}{c^2}\right), \quad x' = \gamma(x - v t), \quad \gamma = \frac{1}{\sqrt{1-\beta^2}}, \ \beta=\frac{v}{c}. $$Use them to test simultaneity ($t'_1=t'_2$) and causal connection (light-cone test: $|\Delta ct| \ge |\Delta x|$).

## Applications- Decide whether two events can be causally connected.- Find the speed that makes two events simultaneous in another frame.

## Key formulas and concepts- Simultaneity condition gives $v = c^2\,\Delta t/\Delta x$ (use $ct$ units so $\beta = \Delta(ct)/\Delta x$ and require $|\beta|\le1$).

```python
from math import sqrtevents = {"A": (3.0,2.0), "B": (9.0,0.5), "C": (5.5,5.0), "D": (10.0,2.5)}  # (ct, x)def lightcone_ok(E, D):    dct = events[D][0]-events[E][0]    dx  = events[D][1]-events[E][1]    return (dct>0) and (dct**2 - dx**2 >= 0)print("A→D causal?", lightcone_ok("A","D"))print("B→D causal?", lightcone_ok("B","D"))def beta_for_sim(E1, E2):    dct = events[E2][0]-events[E1][0]    dx  = events[E2][1]-events[E1][1]    if abs(dx) < 1e-12: return float("inf")    return dct/dxfor lab in ["B","C","D"]:    print("β for A &", lab, "simultaneous:", beta_for_sim("A", lab))
```
