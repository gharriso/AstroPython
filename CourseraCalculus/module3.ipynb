# Limits

What does the function _want_ to do as it approaches some point which is possibly not defined for the function

*Good example is functions that need to avoid divide by zero*

This is determined by taking the x as close to the required value from either size without actually hitting that value




```python
import sympy

# define sympy expression for sin(x)/x 
x= sympy.symbols('x')
expr = sympy.sin(x) / x
# Find the limit for x=0
limit_at_zero = sympy.limit(expr, x, 0)
# Print the limit
print(f"The limit of sin(x)/x as x approaches 0 is: {limit_at_zero}")
# plot the expression
sympy.plot(expr, (x, -10, 10), title='Plot of sin(x)/x', xlabel='x', ylabel='sin(x)/x', show=True)
 

 
    
```

## There can only be one limit

There is no official limit for this function, though sympy does try....

```python
x= sympy.symbols('x')
expr = sympy.sin(sympy.pi/x )   
# Find the limit for x=0
limit_at_zero = sympy.limit(expr, x, 0)
#plot the function
print(f"The limit of sin(pi/x) as x approaches 0 is: {limit_at_zero}")
sympy.plot(expr, (x, -10, 10), title='Plot of sin(pi/x)', xlabel='x', ylabel='sin(pi/x)', show=True)
```

```python
import sympy
x=sympy.symbols('x')
expr = 1/x**2
print("The limit of 1/x^2 as x approaches 0 is: ",sympy.limit(expr, x, 0))
# Find the limit when x is infinite
print("The limit of 1/x^2 as x approaches infinity is: ", sympy.limit(expr, x, sympy.oo))
# plot the expression
sympy.plot(expr, (x, -2, 2), title='Plot of 1/x^2', xlabel='x', ylabel='1/x^2', ylim=(0, 1000), show=True)
```
