```python
from sympy import *
p = symbols('p', integer=True)
n = symbols('n', real=True)
a, b = symbols('a b', positive=True)
from sympy import powsimp
e1 = a**p * a**n
e1a =powsimp(e1)
display(e1,e1a)
print('---------------------')

from sympy import expand_power_exp
e2 = (a**b)**(p + n)
display(e2)
e2a=expand_power_exp(e2)
display(e2)
print('---------------------')

from sympy import expand_power_base
e3 = (a*b)**(p + n)
display(e3)
e3a = expand_power_base(e3)
display(e3a)
print('---------------------')



from sympy import powdenest
e4 = ((a**p)**n)**b
e4a=powdenest(e4)
display(e4,e4a)
print('---------------------')
```
