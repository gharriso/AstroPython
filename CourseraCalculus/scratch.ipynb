```python
from sympy import symbols

def linear(a, b, c, d):
    gradient= (d - b) / (c - a)
    intercept = b - gradient * a
    m,b,x = symbols('m b x')
    expression= m*x + b
    expression.subs({m: gradient, b: intercept})
    return expression.subs({m: gradient, b: intercept})

linear(1,-4,3,6)
```
