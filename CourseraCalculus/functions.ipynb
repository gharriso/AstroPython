```python
from sympy import * 
init_printing(use_unicode=True)
x, y, a, b, c = symbols('x y a b c')
parabolicFunc = a*x**2 + b*x + c
display(parabolicFunc)
# assign values to a, b, and c for plotting
parabolicFunc = parabolicFunc.subs({a: 2, b: 2, c: 10})

from sympy.plotting import plot
# add the function description to the plot title
title = 'Plot of '+parabolicFunc.__str__().replace('**', '^')

plot(parabolicFunc, (x, 0, 10), title=title, show=True)
```

Define a function linear(a, b, c, d), which, when given the coordinates of two points 
(a,b)
 and 
(c,d)
returns the equation of the line passing through those two points.

```python
def linear(a, b, c, d):
    gradient= (d - b) / (c - a)
    intercept = b - gradient * a
    m,b = symbols('m b')
    expression= m*x + b
    expression.subs({m: gradient, b: intercept})
    return expression.subs({m: gradient, b: intercept})

linear(1,2,2,5)
```

```python
# Implement a piecewise defined function in SymPy
from sympy import Piecewise
from sympy import symbols 
def piecewiseFunc(x):
    return Piecewise((x**2, x < 0), (x + 1, x >= 0))
x=symbols('x')
display(piecewiseFunc(x))
```
