# 📘 Differentiation Rules Cheat Sheet

This sheet shows key differentiation rules in both **prime notation** (e.g., `f'(x)`) and **Leibniz notation** (e.g., `df/dx`).

---

## 🔹 1. Power Rule

If  
\[
f(x) = x^n
\]  
then  
\[
f'(x) = n \cdot x^{n-1} \quad \text{or} \quad \frac{d}{dx}x^n = n x^{n-1}
\]

---

## 🔹 2. Constant Rule

\[
f(x) = c \Rightarrow f'(x) = 0 \quad \text{or} \quad \frac{d}{dx}(c) = 0
\]

---

## 🔹 3. Constant Multiple Rule

\[
\frac{d}{dx}(c \cdot f(x)) = c \cdot \frac{d}{dx}f(x) \quad \text{or} \quad (c f(x))' = c f'(x)
\]

---

## 🔹 4. Sum and Difference Rule

\[
\frac{d}{dx}(f(x) \pm g(x)) = f'(x) \pm g'(x)
\]

---

## 🔹 5. Product Rule

\[
\frac{d}{dx}[u(x) \cdot v(x)] = u'(x) v(x) + u(x) v'(x) \\
\text{or} \quad \frac{d}{dx}(uv) = \frac{du}{dx} v + u \frac{dv}{dx}
\]

---

## 🔹 6. Quotient Rule

\[
\frac{d}{dx}\left[\frac{u(x)}{v(x)}\right] = \frac{u'(x) v(x) - u(x) v'(x)}{[v(x)]^2} \\
\text{or} \quad \frac{d}{dx}\left(\frac{u}{v}\right) = \frac{v \frac{du}{dx} - u \frac{dv}{dx}}{v^2}
\]

---

## 🔹 7. Chain Rule

\[
\frac{d}{dx}f(g(x)) = f'(g(x)) \cdot g'(x) \\
\text{or} \quad \frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}
\]

---

## 🔹 8. Exponential and Logarithmic Functions

\[
\frac{d}{dx}(e^x) = e^x \\
\frac{d}{dx}(a^x) = a^x \ln a \\
\frac{d}{dx}(\ln x) = \frac{1}{x}
\]

---

## 🔹 9. Trigonometric Functions

\[
\frac{d}{dx}(\sin x) = \cos x \\
\frac{d}{dx}(\cos x) = -\sin x \\
\frac{d}{dx}(\tan x) = \sec^2 x
\]
