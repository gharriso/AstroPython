```python
import sympy as sp

# Define the variable
t = sp.symbols('t')

# Define inner and outer functions
inner = 3*t**2
outer = 6*sp.sin(inner)

# Differentiate automatically with SymPy
dy_dt_auto = sp.diff(outer, t)

# Manual chain rule steps
# 1. Differentiate outer w.r.t inner
d_outer_d_inner = sp.diff(6*sp.sin(sp.Symbol('u')), sp.Symbol('u')).subs(sp.Symbol('u'), inner)

# 2. Differentiate inner w.r.t t
d_inner_dt = sp.diff(inner, t)

# 3. Multiply them (Chain Rule)
dy_dt_manual = sp.simplify(d_outer_d_inner * d_inner_dt)

# Show results
print("Automatic differentiation:", dy_dt_auto)
print("Manual chain rule:", dy_dt_manual)
print("Are they the same?", sp.simplify(dy_dt_auto - dy_dt_manual) == 0)
```

# Max and Min values

At the max or minimum point, the gradient of the tangent line will be 0



*The critical numbers of f(x) are where f'(x)=0 or f'(x) = DNE*

##  Example

Let’s say a planet’s distance from a star varies with time as:

$$
r(t) = a\bigl(1 - e\cos(\omega t)\bigr)
$$

Where:
- $a$ is semi‑major axis  
- $e$ is eccentricity  
- $\omega$ is angular frequency  

👉 **Task:** Find the max and minimum points

```python
import sympy as sp
t, a, e, omega = sp.symbols('t a e omega', positive=True, real=True)

r = a*(1 - e*sp.cos(omega*t))
drdt = sp.diff(r, t)
#
# Use sympy to find where the differential is zero
#
crtital_times = sp.solve(sp.Eq(drdt, 0), t)              # general: t = n*pi/omega

for critical_time in crtital_times:
    display(critical_time)
    
# This tells us that the differntial is 0 at t=pi/omega

# pick the first maximum: t = pi/omega
 
 
r_max = sp.simplify(r.subs(t, critical_time))                 # -> a*(1 + e)

r_max, critical_time 
```
