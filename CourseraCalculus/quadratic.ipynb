```python
from numpy import sqrt
from sympy.plotting import *
from sympy import Symbol
#ax^2 +bx +c

def zeros(a,b,c):
    D=sqrt(b*b-4*a*c)
    x1=(-b +D)/(2*a)
    x2=(-b-D)/(2*a)
    print("First second root",x1,x2)
    
if __name__ == "__main__":
    print("Quadratic equation solver")
    a=input("Enter a:  ")
    b=input("Enter b:  ")
    c=input("Enter c:  ")
    zeros(float(a),float(b),float(c))
    
```
